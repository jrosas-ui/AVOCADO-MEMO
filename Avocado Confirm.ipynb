```python
# Import Packages
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
```

```python
# Read the 2025 dataset and store it as data2025
data2025=pd.read_csv('2025.csv', encoding='utf-8')

# Read the 2026 dataset and store it as data2026
data2026=pd.read_csv('2026PartYear.csv', encoding='utf-8')
```

```python
for column in data2025.columns:
    print(column)
```

```python
# Total U.S. volume for Weeks 1-32

volume2025=data2025[
    (data2025['Geography']=='Total U.S.') &
    (data2025['Period']<=32)
]['Total Bulk and Bags'].sum()

volume2026=data2026[
    (data2026['Geography']=='Total U.S.') &
    (data2026['Period']<=32)
]['Total Bulk and Bags'].sum()

print("2025 Volume:",volume2025)
print("2026 Volume:",volume2026)
```

```python
# Keep Weeks 1-32 for both years
data2025_32 = data2025[data2025['Period'] <= 32].copy()
data2026_32 = data2026[data2026['Period'] <= 32].copy()

print("2025 Weeks:", data2025_32['Period'].min(), "-", data2025_32['Period'].max())
print("2026 Weeks:", data2026_32['Period'].min(), "-", data2026_32['Period'].max())
```

```python
# Keep only Total U.S. records for national calculations
us2025 = data2025_32[
    data2025_32['Geography'] == 'Total U.S.'
].copy()

us2026 = data2026_32[
    data2026_32['Geography'] == 'Total U.S.'
].copy()
```

## Total U.S. Volume

```python
# Total U.S. volume for Weeks 1-32

volume2025 = us2025['Total Bulk and Bags'].sum()
volume2026 = us2026['Total Bulk and Bags'].sum()

print("2025 Volume:", volume2025)
print("2026 Volume:", volume2026)
```

```python
# Calculate percentage change in Total U.S. volume

volume_change = (
    (volume2026 - volume2025) /
    volume2025
) * 100

print("Volume Change:", round(volume_change, 1), "%")
```

## Average Selling Price (ASP)

```python
# Calculate revenue for each Total U.S. record

us2025['Calculated Revenue'] = (
    us2025['ASP Current Year'] *
    us2025['Total Bulk and Bags']
)

us2026['Calculated Revenue'] = (
    us2026['ASP Current Year'] *
    us2026['Total Bulk and Bags']
)
```

```python
# Calculate weighted average ASP for Total U.S.

asp2025 = (
    us2025['Calculated Revenue'].sum() /
    us2025['Total Bulk and Bags'].sum()
)

asp2026 = (
    us2026['Calculated Revenue'].sum() /
    us2026['Total Bulk and Bags'].sum()
)

print("2025 ASP:", round(asp2025, 3))
print("2026 ASP:", round(asp2026, 3))
```

```python
# Calculate weighted average ASP for Total U.S.

asp2025 = (
    us2025['Calculated Revenue'].sum() /
    us2025['Total Bulk and Bags'].sum()
)

asp2026 = (
    us2026['Calculated Revenue'].sum() /
    us2026['Total Bulk and Bags'].sum()
)

print("2025 ASP:", round(asp2025, 3))
print("2026 ASP:", round(asp2026, 3))
```

```python
# Calculate percentage change in ASP

asp_change = (
    (asp2026 - asp2025) /
    asp2025
) * 100

print("ASP Change:", round(asp_change, 1), "%")
```

## Conventional and Organic Avocado Volume

```python
# Total volume by Type for 2025

volume_by_type2025 = (
    us2025.groupby('Type')['Total Bulk and Bags']
    .sum()
)

print("2025 Volume by Type")
print(volume_by_type2025)
```

```python
# Total volume by Type for 2026

volume_by_type2026 = (
    us2026.groupby('Type')['Total Bulk and Bags']
    .sum()
)

print("2026 Volume by Type")
print(volume_by_type2026)
```

```python
# Conventional volume change

conventional_change = (
    (volume_by_type2026['Conventional'] -
     volume_by_type2025['Conventional']) /
    volume_by_type2025['Conventional']
) * 100

print(
    "Conventional Volume Change:",
    round(conventional_change, 1),
    "%"
)
```

```python
# Organic volume change

organic_change = (
    (volume_by_type2026['Organic'] -
     volume_by_type2025['Organic']) /
    volume_by_type2025['Organic']
) * 100

print(
    "Organic Volume Change:",
    round(organic_change, 1),
    "%"
)
```

```python
# Calculate weighted ASP by Type

asp_by_type2025 = (
    us2025.groupby('Type')['Calculated Revenue'].sum() /
    us2025.groupby('Type')['Total Bulk and Bags'].sum()
)

asp_by_type2026 = (
    us2026.groupby('Type')['Calculated Revenue'].sum() /
    us2026.groupby('Type')['Total Bulk and Bags'].sum()
)

print("2025 ASP by Type")
print(asp_by_type2025.round(3))

print("\n2026 ASP by Type")
print(asp_by_type2026.round(3))
```

```python
# Calculate ASP percentage changes by Type

conventional_asp_change = (
    (asp_by_type2026['Conventional'] -
     asp_by_type2025['Conventional']) /
    asp_by_type2025['Conventional']
) * 100

organic_asp_change = (
    (asp_by_type2026['Organic'] -
     asp_by_type2025['Organic']) /
    asp_by_type2025['Organic']
) * 100

print(
    "Conventional ASP Change:",
    round(conventional_asp_change, 1),
    "%"
)

print(
    "Organic ASP Change:",
    round(organic_asp_change, 1),
    "%"
)
```

```python
# Calculate Conventional and Organic volume shares

conventional_share2025 = (
    volume_by_type2025['Conventional'] /
    volume2025
) * 100

organic_share2025 = (
    volume_by_type2025['Organic'] /
    volume2025
) * 100

conventional_share2026 = (
    volume_by_type2026['Conventional'] /
    volume2026
) * 100

organic_share2026 = (
    volume_by_type2026['Organic'] /
    volume2026
) * 100

print("2025 Conventional Share:", round(conventional_share2025, 1), "%")
print("2025 Organic Share:", round(organic_share2025, 1), "%")
print("2026 Conventional Share:", round(conventional_share2026, 1), "%")
print("2026 Organic Share:", round(organic_share2026, 1), "%")
```

## Volume by Avocado Variety

```python
# Total volume by variety for 2025

variety2025 = us2025[['4046', '4225', '4770']].sum()

print("2025 Volume by Variety")
print(variety2025)
```

```python
# Total volume by variety for 2026

variety2026 = us2026[['4046', '4225', '4770']].sum()

print("2026 Volume by Variety")
print(variety2026)
```

## Regional Volume Changes

The dataset contains eight regional geographies:
California, Great Lakes, Midsouth, Northeast, Plains, South Central, Southeast, and West.

```python
# Regional geographies included in the dataset

regions = [
    'California',
    'Great Lakes',
    'Midsouth',
    'Northeast',
    'Plains',
    'South Central',
    'Southeast',
    'West'
]
```

```python
# Calculate total volume by region for Weeks 1-32

regional2025 = (
    data2025_32[
        data2025_32['Geography'].isin(regions)
    ]
    .groupby('Geography')['Total Bulk and Bags']
    .sum()
)

regional2026 = (
    data2026_32[
        data2026_32['Geography'].isin(regions)
    ]
    .groupby('Geography')['Total Bulk and Bags']
    .sum()
)

print("2025 Regional Volume")
print(regional2025)
```

```python
print("2026 Regional Volume")
print(regional2026)
```

```python
# Calculate percentage change by region

regional_change = (
    (regional2026 - regional2025) /
    regional2025
) * 100

print("Regional Volume Percentage Change")
print(regional_change.round(1))
```

```python
# Calculate weighted ASP by region

regional_asp2025 = (
    data2025_32[
        data2025_32['Geography'].isin(regions)
    ]
    .assign(
        CalculatedRevenue=lambda x:
        x['ASP Current Year'] * x['Total Bulk and Bags']
    )
    .groupby('Geography')['CalculatedRevenue']
    .sum()
    /
    data2025_32[
        data2025_32['Geography'].isin(regions)
    ]
    .groupby('Geography')['Total Bulk and Bags']
    .sum()
)

regional_asp2026 = (
    data2026_32[
        data2026_32['Geography'].isin(regions)
    ]
    .assign(
        CalculatedRevenue=lambda x:
        x['ASP Current Year'] * x['Total Bulk and Bags']
    )
    .groupby('Geography')['CalculatedRevenue']
    .sum()
    /
    data2026_32[
        data2026_32['Geography'].isin(regions)
    ]
    .groupby('Geography')['Total Bulk and Bags']
    .sum()
)

print("2025 Regional ASP")
print(regional_asp2025.round(3))

print("\n2026 Regional ASP")
print(regional_asp2026.round(3))
```

```python
# Calculate regional ASP percentage changes

regional_asp_change = (
    (regional_asp2026 - regional_asp2025) /
    regional_asp2025
) * 100

print("Regional ASP Percentage Change")
print(regional_asp_change.round(1))
```

## Bagged Avocado Volume

```python
# Total bagged volume for 2025 and 2026

bagged2025 = us2025['TotalBagged'].sum()
bagged2026 = us2026['TotalBagged'].sum()

print("2025 Bagged Volume:", bagged2025)
print("2026 Bagged Volume:", bagged2026)
```

```python
# Calculate percentage change in total bagged volume

bagged_change = (
    (bagged2026 - bagged2025) /
    bagged2025
) * 100

print("Bagged Volume Change:", round(bagged_change, 1), "%")
```

```python
# Calculate changes in bagged size categories

bagged_sizes = [
    'Sml Bagged',
    'Lrg Bagged',
    'X-Lrg Bagged',
    'Unknown Bag Size'
]

for size in bagged_sizes:
    value2025 = us2025[size].sum()
    value2026 = us2026[size].sum()

    change = (
        (value2026 - value2025) /
        value2025
    ) * 100

    print(size, "Change:", round(change, 1), "%")
```
